**Meta1 dictionary and the resulting metadata_df DataFrame**

```python
# Create metadata Meta1, based on Stanley Smith Steven's definition of variables: Binary, Nominal, Interval, Ordinal.
Meta1 = {
    'Binary': [],  # Replace with actual binary column names
    'Nominal': [], # Replace with actual nominal column names
    'Ordinal': [], # Replace with actual ordinal column names
    'Interval': [] # Replace with actual interval column names
}
# Categorize columns based on their data types
for column in train.columns:
    # Check if it's binary (0/1 or True/False)
    if train[column].dtype in ['int64', 'bool']:
        if train[column].nunique() == 2:  # Check for two unique values
            Meta1['Binary'].append(column)  # Add to Binary
        else:
            Meta1['Interval'].append(column)  # If not binary, treat as Interval
    elif train[column].dtype == 'float64':
        Meta1['Interval'].append(column)  # Add float64 to Interval
    elif train[column].dtype == 'object':
        Meta1['Nominal'].append(column)  # Add string/object to Nominal
    # Check if it's an ordinal variable (customize the condition as needed)
    elif train[column].dtype == 'int64' and column in ['satisfaction', 'rating']:  # Example for ordinal
        Meta1['Ordinal'].append(column)

# Convert the metadata into a DataFrame
metadata_df = pd.DataFrame.from_dict(Meta1, orient='index').transpose()

# Display the metadata DataFrame
print("Variable Metadata DataFrame:")
print(metadata_df)
```

**Observation:**

**Binary:** The Survived column is binary, indicating whether a passenger survived (1) or not (0).

**Nominal:** Columns like Name, Sex, Ticket, and Cabin are nominal (categorical data without any intrinsic order). Embarked also falls under this category, as it represents the port where passengers boarded the ship.

**Ordinal:** There are no columns marked as ordinal, meaning there are no categorical variables with a specific order (e.g., rating scales).

**Interval:** Columns such as PassengerId, Pclass, Age, SibSp, Parch, and Fare are interval data (quantitative values that have meaningful differences but no true zero point).

```python
def bar_chart_stacked(train, feature, stacked = True):
    survived = train[train['Survived']==1][feature].value_counts()
    dead = train[train['Survived']==0][feature].value_counts()
    df_survived_dead = pd.DataFrame([survived,dead])
    df_survived_dead.index = ['Passengers Survived','Passengers Died']
    ax = df_survived_dead.plot(kind='bar',stacked=stacked, figsize=(5,5))
```

```python
def bar_chart_compare(train, feature1, feature2=None, title = "Survival rate by sex and class'"):
    plt.figure(figsize = [5,5])
    plt.title(title)
    g = sns.barplot(x=feature1, y='Survived', hue=feature2, ci=None, data=train).set_ylabel('Survival rate')
```

```python
def show_countplot(train, feature, title, fsize = (5,5)):
    fig, ax = plt.subplots(figsize=fsize)
    sns.countplot(train[feature], ax=ax).set_title(title)
    
def show_compare_countplot(train, feature1, feature2, title):
    fig, ax = plt.subplots(figsize=(5,5))
    p = sns.countplot(x = feature1, hue = feature2, data = train, ax=ax).set_title(title)  
```

```python
def plot_distribution(train, feature, title, bins = 30, hist = True, fsize = (5,5)):
    fig, ax = plt.subplots(figsize=fsize)
    ax.set_title(title)
    sns.distplot(train[feature], color='g', bins=bins, ax=ax)
```

```python
def plot_kernel_density_estimate_survivors(train, feature1, title, fsize = (5,5)):
    fig, ax = plt.subplots(figsize=fsize)
    ax.set_title(title) 
    sns.kdeplot(train[feature1].loc[train["Survived"] == 1],
                shade= True, ax=ax, label='Survived').set_xlabel(feature1)
    sns.kdeplot(train[feature1].loc[train["Survived"] == 0],
                shade=True, ax=ax, label="Died")
```

```python
def plot_quartiles(train, feature, title, categories):
    fig, axarr = plt.subplots(figsize=(5,5))
    fare_ranges = pd.qcut(train[feature], len(categories), labels = categories) #. [0, .25, .5, .75, 1.]
    axarr.set_title(title)
    sns.barplot(x=fare_ranges, y=train.Survived, ci=None, ax=axarr).set_ylabel('Survival rate')
```

## Feature Importance Analysis

### Feature Importance for SVC

Support Vector Machines (SVM) do not inherently provide feature importance as tree-based models do. However, if you use a linear kernel (kernel='linear'), you can derive feature importance from the absolute values of the model's coefficients.


**Train the Linear Kernel SVC**

```python
# Train SVM with linear kernel
svc_linear = SVC(kernel='linear', class_weight='balanced', probability=True, random_state=42)
svc_linear.fit(X_train_scaled, y_train)

# Get feature importance (absolute values of coefficients)
feature_importance = pd.DataFrame({
    'Feature': X_train.columns,
    'Importance': abs(svc_linear.coef_[0])
}).sort_values(by='Importance', ascending=False)

print("Feature Importance:")
print(feature_importance)
```

**Plot Feature Importance**

```python
# Plot feature importance
plt.figure(figsize=(10, 6))
plt.barh(feature_importance['Feature'], feature_importance['Importance'])
plt.gca().invert_yaxis()
plt.title('Feature Importance from Linear SVC')
plt.xlabel('Importance')
plt.ylabel('Features')
plt.show()
```

**Retrain the Model with Top Features**

```python
# Select top features
top_features = feature_importance['Feature'].iloc[:10]

# Subset training and testing data
X_train_top = X_train_scaled[:, :len(top_features)]
X_test_top = X_test_scaled[:, :len(top_features)]

# Train SVM with top features
svc_top = SVC(kernel='linear', class_weight='balanced', probability=True, random_state=42)
svc_top.fit(X_train_top, y_train)

# Predict and evaluate the model
y_pred_top = svc_top.predict(X_test_top)
accuracy_top = accuracy_score(y_test, y_pred_top)
print(f"SVM Accuracy with Top Features: {accuracy_top:.4f}")
print("Classification Report:")
print(classification_report(y_test, y_pred_top))
```
